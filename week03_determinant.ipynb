```python
import numpy as np

theta = np.radians(30)
R  = np.array([[np.cos(theta), -np.sin(theta)],
               [np.sin(theta),  np.cos(theta)]])   # turn 30°
S  = np.array([[1, 1],
               [0, 1]])                            # shear right
Sc = np.array([[2, 0],
               [0, 0.5]])                          # x times 2, y times 0.5
M  = np.array([[-1, 0],
               [ 0, 1]])                           # mirror

def det2(A):
    return A[0, 0] * A[1, 1] - A[0, 1] * A[1, 0]

s = 2
SR = s * R
print("s =", s, "  det:", round(det2(SR), 3), "  s squared:", s ** 2)
```

```python
import matplotlib.pyplot as plt

square = np.array([[0, 1, 1, 0, 0],
                   [0, 0, 1, 1, 0]])
F = np.array([[1, 2],
              [2, 4]])
print("det of F:", det2(F))

flat = F @ square
print(flat)

plt.plot(square[0], square[1], "o-", label="before")
plt.plot(flat[0], flat[1], "o-", label="after F")
plt.gca().set_aspect("equal")
plt.grid(True)
plt.legend()
plt.title("det = 0: squashed onto a line")
plt.show()
```

```python
A = np.array([[2, 1],
              [1, 3]])
end = np.array([5, 10])

start = np.linalg.solve(A, end)
print("start point (x, y):", start)
print("check, A @ start:", A @ start)
print("det of A:", det2(A))
```
