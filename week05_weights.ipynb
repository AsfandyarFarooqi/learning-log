## Tuesday: weighted least squares

```python
A = np.array([[1.0], [1.0]])
l = np.array([4.0, 14.0])
sigma = np.array([2.0, 6.0])
P = np.diag(1 / sigma**2)
N = A.T @ P @ A
n = A.T @ P @ l
x_hat = np.linalg.solve(N, n)
r = l - A @ x_hat
print("P =")
print(np.round(P, 4))
print("weighted mean:", x_hat, "mm above 125 m")
print("plain mean:", np.mean(l), "mm above 125 m")
print("residuals (mm):", r)
print("residuals in sigma units:", r / sigma)
```

```python
A_lev = np.array([[1, 0],
                  [-1, 1],
                  [0, 1]])
l_lev = np.array([2.004, 1.502, 3.510])
length_km = np.array([1.0, 1.0, 4.0])
P_lev = np.diag(1 / length_km)


x_plain = np.linalg.solve(A_lev.T @ A_lev, A_lev.T @ l_lev)
x_w = np.linalg.solve(A_lev.T @ P_lev @ A_lev, A_lev.T @ P_lev @ l_lev)
print("plain    h_P, h_Q =", np.round(x_plain, 5))
print("weighted h_P, h_Q =", np.round(x_w, 5))
print("plain residuals (mm):   ", np.round((l_lev - A_lev @ x_plain) * 1000, 2))
print("weighted residuals (mm):", np.round((l_lev - A_lev @ x_w) * 1000, 2))
```

```python
A_lev = np.array([[1, 0],
                  [-1, 1],
                  [0, 1]])
l_lev = np.array([1.207,0.495,1.708])
length_km = np.array([3, 1.0, 2])
P_lev = np.diag(1 / length_km)

At_P_A=A_lev.T@P_lev@A_lev
At_P_l=A_lev.T@P_lev@l_lev
x_w=np.linalg.inv(At_P_A)@At_P_l
x_plain = np.linalg.solve(A_lev.T @ A_lev, A_lev.T @ l_lev)
print("plain    h_P, h_Q =", np.round(x_plain, 5))
print("weighted h_P, h_Q =", np.round(x_w, 5))
print("weighted residuals (mm):   ", np.round((l_lev - A_lev @ x_w) * 1000, 2))
print("plain residuals (mm):   ",np.round((l_lev - A_lev@x_plain)* 1000, 2))
```

```python
## Saturday: how sure are the heights?
A_lev = np.array([[1, 0],
                  [-1, 1],
                  [0, 1]])
length_km = np.array([1.0, 1.0, 4.0])
P_lev = np.diag(1 / length_km)
sigma0 = 1.0
C_x = sigma0**2 * np.linalg.inv(A_lev.T @ P_lev @ A_lev)
print("C_x (mm^2) =")
print(np.round(C_x, 4))
print("sigma h_P =", round(np.sqrt(C_x[0, 0]), 3), "mm")
print("sigma h_Q =", round(np.sqrt(C_x[1, 1]), 3), "mm")
```
