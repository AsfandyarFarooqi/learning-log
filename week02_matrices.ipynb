```python
import numpy as np

R = np.array([[0, -1],
              [1,  0]])     # turn 90° left
S = np.array([[1, 1],
              [0, 1]])      # slant right

i_hat = np.array([1, 0])

print("first R, then S:", S @ (R @ i_hat))
print("first S, then R:", R @ (S @ i_hat))

print("one matrix for 'R then S':")
print(S @ R)
print("one matrix for 'S then R':")
print(R @ S)
```

```python
#Wednesday transforming a cell
import numpy as np
import matplotlib.pyplot as plt

square = np.array([
    [0, 1, 1, 0, 0],   # x of each corner
    [0, 0, 1, 1, 0],   # y of each corner
])


def show(before, after, title):
    plt.plot(before[0], before[1], "o-", label="before")
    plt.plot(after[0], after[1], "o-", label="after")
    plt.gca().set_aspect("equal")
    plt.grid(True)
    plt.legend()
    plt.title(title)
    plt.show()


theta = np.radians(90)
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print("R =")
print(np.round(R,0),theta)
show(square, R @ square, "rotate 90° left")
```

```python
import numpy as np
import matplotlib.pyplot as plt

square=np.array([[0,1,1,0,0],
                [0,0,1,1,0]])
shear_S=np.array([[1,-1],
                  [0,1]])
theta = np.radians(30)
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
transformed_square=shear_S@square

plt.plot(square[0],square[1],label='Org')
plt.plot(transformed_square[0],transformed_square[1],label='Transformed')
plt.gca().set_aspect("equal")
plt.grid(True)
plt.legend()
plt.title('Shear Transformation')
plt.show()

shift = np.array([[-1],
                  [ 0]])
moved = square + shift
print(moved)

plt.plot(square[0], square[1], label="Org")
plt.plot(moved[0], moved[1], label="Shifted")
plt.gca().set_aspect("equal")
plt.grid(True)
plt.legend()
plt.title("Slide left by 1")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

square=np.array([[0,1,1,0,0],
                [0,0,1,1,0]])
S=np.array([[1,1],
                  [0,1]])
theta = np.radians(30)
R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
SR = S @ R    # first rotate, then shear
RS = R @ S    # first shear, then rotate
print("S @ R =")
print(np.round(SR, 3))
print("R @ S =")
print(np.round(RS, 3))

fig, axes = plt.subplots(1, 2, figsize=(10, 5))
for ax, M, name in zip(axes, [SR, RS], ["S @ R: rotate, then shear", "R @ S: shear, then rotate"]):
    moved = M @ square
    ax.plot(square[0], square[1], "o-", label="before")
    ax.plot(moved[0], moved[1], "o-", label="after")
    ax.set_aspect("equal")
    ax.grid(True)
    ax.legend()
    ax.set_title(name)
plt.show()
```
