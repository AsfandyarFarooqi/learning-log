```python
import numpy as np

R = np.array([[0, -1],
              [1,  0]])     # turn 90° left
S = np.array([[1, 1],
              [0, 1]])      # slant right

i_hat = np.array([1, 0])

print("first R, then S:", S @ (R @ i_hat))
print("first S, then R:", R @ (S @ i_hat))

print("one matrix for 'R then S':")
print(S @ R)
print("one matrix for 'S then R':")
print(R @ S)
```
